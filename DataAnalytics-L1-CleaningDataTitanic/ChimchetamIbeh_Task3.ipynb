```python
#Importing Libraries
import pandas as pd
import numpy as np
```

```python
#Loading the dataset
df = pd.read_csv("C:/Users/Admin/Documents/My Data Sources/Titanic-Dataset.csv")
```

```python
#initial inspection
print ("DATA QUALITY REPORT\n")
print(f"Shape:{df.shape}\n")
print(f"Duplicate rows:\n{df.duplicated().sum()}\n")
print(f"Dtypes:\n{df.dtypes}\n")
print(f"Nulls:\n{df.isnull().sum()}\n")
print(f"Summary statistics:\n{df.describe()}")
```

**OBSERVATIONS:**  
The dataset contains 891 rows and 12 columns.  
**Nulls:** The age column has 177 empty cells, the cabin column has 687 empty cells, amd the Embarked column has only 2 empty cells.  
**Duplicates:** There are no duplicated rows.  
**Dtypes:** All columns are appropriately matched.  
**Value Range:** Age ranges from 0.42(children less than 1 year old) to 80. Fare ranges from 0.0 to 512.33.

```python
df.head()
```

```python
#creating a copy of the datset
df_clean = df.copy()
```

```python
#Handling Missing Data (Age)

#Extracting titles
df_clean['Title'] = df_clean['Name'].str.split(',').str[1].str.split('.').str[0].str.strip()

#Confirming output
print (df_clean['Title'].value_counts())

#Filling Null Age cells with median age
df_clean['Age'] = df_clean.groupby('Title')['Age'].transform(lambda x:x.fillna(x.median()))

#Confirming Correction Implementation
print (f"\nNull_Age: {df_clean['Age'].isnull().sum()}\n")
```

The **Age** column had 177 missing values. Rather than filling with an overall median, **age** was grouped by title and the missing values were filled with median age within each title group, giving a more reliable estimate.

```python
#Handling Missing Data (Cabin)
df_clean['Has_Cabin'] = df_clean['Cabin'].apply(lambda x: 1 if pd.notnull (x)
                                                else 0)

#Confirming Correction Implementation
print(df_clean['Has_Cabin'].value_counts())

#Dropping Cabin column
df_clean.drop(columns=['Cabin'], inplace = True)

#Confirming Correction Implementation
df_clean.head()

```

The **Cabin** had a total of *687* missing values, i assumed these represented passengers without cabins. A new column named **Has_Cabin** was created, with all empty cells classified as **0** representing passengers *without* designated cabins and all others classified as *1*  representing passengers who were assigned cabins.  
I used binary so it corresponds with other variables in the dataset; to enhance uniformity.

```python
#Handling Missing Data (Embarked)

#Filling Null cell with U (Unknown)
df_clean['Embarked'] = df_clean['Embarked'].fillna('U')

#Confirming Correction Implementation
print (df_clean['Embarked'].value_counts())
```

The **Embarked** column had *2* missing values. Rather than guessing and filling these missing values with a specific port, the empty cells were filled using *U* to represent *Unknown*, to enhance uniformity. 

```python
#Duplicate rows removal
duplicate_count = df_clean.duplicated().sum()
print(f"Number of Duplicates Found: {duplicate_count}")
```

This dataset has no duplicates hence zero duplicates were removed.

```python
#Standardization

#Checking for inconsistencies
print(f"{df_clean['Sex'].value_counts()}\n")
print(f"{df_clean['Title'].value_counts()}\n")
print(f"{df_clean['Embarked'].value_counts()}")

#Changing sex cells to title case
df_clean['Sex'] = df_clean['Sex'].apply(lambda x: x.title() if isinstance(x, str) else x)

#Confirming correction implementation
print(f"\nCorrected:\n{df_clean['Sex'].value_counts()}")
```

There wasn't any major case of inconsistency within any column. The values in the sex column was changed to Title case just to correspond with values in other column within the dataset, so everything remains has a uniform appearance.

```python
#Outlier detection (Fare)
#using Zscore
Mean_Fare = df_clean['Fare'].mean()
Std_Fare = df_clean['Fare'].std()

df_clean['Fare_Zscore'] = (df_clean['Fare'] - Mean_Fare)/ Std_Fare

Outliers = df_clean[df_clean['Fare_Zscore'].abs() > 3]
print(f"Number of Fare outliers: {len(Outliers)}")
```

```python
#Verification Check
Outliers.sort_values('Fare', ascending=False)
```

```python
#Checking total number of 1st class passengers
First_class_passengers = df_clean[df_clean['Pclass'] == 1]
print(f"Total First Class Passengers: {len(First_class_passengers)}")

#Getting Fare distribution among 1st class passengers, excluding outliers
First_class_fares = First_class_passengers[First_class_passengers['Fare_Zscore'].abs() <= 3]['Fare']

#Creating a maximum normal 1st class fare cap
fare_cap = First_class_fares.max()
print(f"Fare_Cap: {fare_cap:.2f}")

#Applying the Fare_cap to only outlier rows
df_clean.loc[df_clean['Fare_Zscore'].abs() > 3,'Fare'] = df_clean.loc[df_clean['Fare_Zscore'].abs() > 3,'Fare'].clip(upper=fare_cap)
```

```python
#confirming capping was implemented
df_clean['Fare'].describe()
```

**Observation:** The **Fare** column had a total of *20* outliers. A review of these rows showed that they all correspond with First class passengers, reflecting higher Fare cost. These outliers were *capped* to the maximum non-outlier fare observed within First class passengers specifically *(~164.87)*,this preserves realistic pricing, since First class fares are naturally cost more. 

```python
#Outlier detection (Age)
#using Zscore
Mean_Age = df_clean['Age'].mean()
Std_Age = df_clean['Age'].std()

df_clean['Age_Zscore'] = (df_clean['Age'] - Mean_Age)/ Std_Age

Age_outliers = df_clean[df_clean['Age_Zscore'].abs() > 3]
print(f"Number of Age outliers: {len(Age_outliers)}")
```

```python
#Verification Check
Age_outliers.sort_values('Age', ascending=False)
```

**Observation:** Also using the Z-score method, *7* outliers were identified in the **Age** column. These rows corresponded to elderly **passengers** . These outliers were *retained* because capping or removing them would misrepresent actual passenger information.

```python
#Data type correction
print(df_clean.dtypes)
```

**Observation:** As shown in the initial data quality report carried out on the dataset, all columns are already appropriately typed.
